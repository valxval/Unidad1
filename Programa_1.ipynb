# **Tema de listas**

```python
alumnos = ["Ana", "Luis", "Carlos", "Juan"]
print(alumnos)
print(alumnos[0])
print(alumnos[3])
print(alumnos[5]) #Esta instruciion marca error porque la posicion no existe
```

```python
alumnos[1]= " Maria Luisa"
print(alumnos)
```

```python
# El metodo append agrega un elemento al final de la lista
alumnos.append("Karla")
print(alumnos)
```

```python
# El metodo Inser agrega un elemento a lista en la posicion especificada
alumnos.insert(2, "Cristina")
print(alumnos)
```

```python
# remove elimna un elemento especificando el valor del elemento
# debe coincidir exactamente mayusculas y minusculas
alumnos.remove("Ana")
alumnos.remove("ana") # Este marcaria error porque no esta bien escrito
```

```python
print(alumnos)
```

```python
# pop elimina el elemento o item en la posicion especificada
alumnos.pop(2)
print(alumnos)
```

```python
# del tambien elimina el elemento o item en la posicion especificada
del alumnos[3]
print(alumnos)
```

```python
# La funcion len me permite conocer cuantos elementos tiene una lista
print("Elementos en la lista de alumnos", len(alumnos))
```
